```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Ajuste o caminho do arquivo Excel
arquivo_excel = "dados.xlsx"

# Leitura da planilha
df = pd.read_excel(arquivo_excel)

# Configurações visuais
sns.set(style="whitegrid")
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

# Visão geral
print("Dimensão da base:", df.shape)
display(df.head())
print("\nTipos de dados:")
print(df.dtypes)

print("\nResumo estatístico:")
display(df.describe(include="all").T)

print("\nValores ausentes por coluna:")
display(df.isna().sum().sort_values(ascending=False))

print("\nLinhas duplicadas:", df.duplicated().sum())

# Separação de colunas numéricas e categóricas
colunas_numericas = df.select_dtypes(include="number").columns
colunas_categoricas = df.select_dtypes(exclude="number").columns

# Análise univariada numérica
if len(colunas_numericas) > 0:
    df[colunas_numericas].hist(figsize=(14, 10), bins=20, edgecolor="black")
    plt.suptitle("Distribuição das variáveis numéricas", y=1.02)
    plt.tight_layout()
    plt.show()

# Análise univariada categórica
for coluna in colunas_categoricas[:10]:
    plt.figure(figsize=(10, 4))
    df[coluna].value_counts(dropna=False).head(10).plot(kind="bar")
    plt.title(f"Distribuição de {coluna}")
    plt.xlabel(coluna)
    plt.ylabel("Contagem")
    plt.tight_layout()
    plt.show()

# Correlação entre variáveis numéricas
if len(colunas_numericas) > 1:
    plt.figure(figsize=(10, 8))
    sns.heatmap(df[colunas_numericas].corr(), annot=True, cmap="coolwarm", fmt=".2f")
    plt.title("Mapa de correlação")
    plt.tight_layout()
    plt.show()

# Heatmap de valores ausentes
plt.figure(figsize=(12, 6))
sns.heatmap(df.isna(), cbar=False, cmap="viridis")
plt.title("Mapa de calor de valores ausentes")
plt.xlabel("Colunas")
plt.ylabel("Observações")
plt.tight_layout()
plt.show()
```
